```python

from scipy import stats
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.stats.weightstats import DescrStatsW
# The central limit theorem
dist = stats.uniform(0, 1)
# suppose, that we know the mean and standard deviation
z = dist.mean()
sigma = dist.std() # standard deviation for uniform(0, 1) distribution, which is
print(z)
print(sigma)
# show the central limit theorem
def cltSim(n, nruns, dist):
    Ysim = np.zeros(nruns)
    for i in range(nruns) :
        Zsim = dist.rvs(n) # n random numbers
        z = dist.mean() # theoretical mean
        sigma = dist.std() # theoretical standard deviation
        Yn = np.sqrt(n)*(np.mean(Zsim)-z)/sigma
# note that in the above line, according to Slutsky's Theorem,
# we can replace sigma by std(Zsim)
        Ysim[i] = Yn
        return Ysim
# plot the estimated CDF and PDF, and compare it to the
# standard normal distribution
normDist = stats.norm(0, 1)
n = 10
sims = cltSim(n, 100000, dist)
plt.figure()
plt.hist(sims, bins=100, rwidth=0.8, density=True)
x = np.arange(-3, 3, 0.1)
plt.plot(x, normDist.pdf(x), color="red")
plt.show()
# Confidence interval
# Example 1: for the mean of a geometric distribution
n = 1000000
p = 0.8
p2 = 1-p # careful: in Python the geometric distribution is defined with parameter
#'1-p' compared to the lecture notes
sim = stats.geom(loc=-1, p=p2).rvs(n)
# Theoretical mean: p/(1-p) = 4
# Theoretical variance: p/(1-p)^2 = 20
print(np.mean(sim),np.var(sim))
# confidence interval for the mean
# 1. using the theoretical variance
s2 = p/(1-p)**2
m = np.mean(sim)
z = 1.96
halfWidth = z * np.sqrt(s2/n)
interval = (m - halfWidth, m + halfWidth)
print(interval)
# 2. Using Slutsky's theorem: estimate the sample variance
s2 = np.var(sim)
m = np.mean(sim)
z = 1.96
halfWidth = z * np.sqrt(s2 / n)
interval = (m - halfWidth, m + halfWidth)
print(interval)
# Alternative, using statsmodels
from statsmodels.stats.weightstats import DescrStatsW
ci = DescrStatsW(sim).tconfint_mean(alpha=0.05)
print(ci)
```
